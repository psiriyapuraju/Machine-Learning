```python
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
iris
```

```python
X = iris.data
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
model = GaussianMixture(n_components = 3, random_state = 42)
X_label = model.fit_predict(X_scaled)
```

```python
print("Cluster labels:\n", X_label)
print("Cluster probabilities:\n", model.predict_proba(X_scaled))
```

```python
print("Silhouette Score:", silhouette_score(X_scaled, X_label))
print("Calinski Harabasz Score:", calinski_harabasz_score(X_scaled, X_label))
print("Davies Bouldin Score:", davies_bouldin_score(X_scaled, X_label))
print("Means:\n", model.means_)
print("Covariances:\n", model.covariances_)
print("Weights:\n", model.weights_)
```
